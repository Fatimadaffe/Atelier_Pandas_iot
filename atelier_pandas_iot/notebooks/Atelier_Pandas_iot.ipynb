Partie 5 – Manipulation des colonnes

```python
#1) Ajouter la colonne "temperature_fahrenheit"
df["temperature_fahrenheit"] = (
    df["temperature"] * 9 / 5 + 32
)
df.head()
#2) Ajouter la colonne "niveau_temperature" pour catégoriser la température en "Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy
df["niveau_temperature"] = np.where(
    df["temperature"] > 30,
    "Élevée",
    "Normale"
)
df[["temperature", "niveau_temperature"]].head()
#3) Renommer la colonne "humidite" en "humidite_relative"
df.rename(
    columns={"humidite": "humidite_relative"},
    inplace=True
)
#4) Supprimer la colonne "temperature_fahrenheit"
df.drop(
    columns=["temperature_fahrenheit"],
    inplace=True
)
```

Partie 6 – Filtrage

```python
#1) Récupérer uniquement les lignes telles que température > 30
df_temperature_elevee = df[
    df["temperature"] > 30
]

print(df_temperature_elevee)
#2) Récupérer uniquement les lignes telles que température > 30 et humidité > 70
df_filtre = df[
    (df["temperature"] > 30) &
    (df["humidite_relative"] > 70)
]

print(df_filtre)
```
